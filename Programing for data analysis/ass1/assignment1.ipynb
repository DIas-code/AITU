### 2. Работа с CSV

- В папку `data` сохранить файл `students.csv` с данными о студентах (ФИО, группа, оценки).
- Считать данные из `students.csv` и вычислить средний балл по каждому студенту.

```python
import csv
path_to_file = os.path.join('project/data/', 'students.csv')

with open(path_to_file, 'r', encoding='utf-8') as file:
    reader = csv.DictReader(file)
    for student in reader:
        average = ((int(student['math'])+int(student['physics']) + int(student['cs']) + 
                   int(student['history']) + int(student['english']))/5)

        print(student['full_name'], round(average, 2))
        
```

### 4. Архивация

- Автоматически создать архив `results.zip`, в котором сохраняется папка `results`.

```python
#3&4
import csv
import json
import zipfile

path_to_students = os.path.join('project/data/', 'students.csv')
path_to_report = os.path.join('project/results/', 'report.json')

with open(path_to_report, 'w', encoding='utf-8') as report:
    with open(path_to_students, 'r', encoding='utf-8') as students_file:
        reader = csv.DictReader(students_file)
        for student in reader:
            average = ((int(student['math'])+int(student['physics']) + int(student['cs']) + 
                    int(student['history']) + int(student['english']))/5)
            result = {
                        'full_name': student['full_name'],
                        'average': average
                    }
            json.dump(result, report, indent=4)

with zipfile.ZipFile('project/results.zip', 'w', compression=zipfile.ZIP_DEFLATED) as archive:
    archive.write(
        'project/results/report.json',
        'results/report.json'
    )
```
