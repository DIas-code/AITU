The corpus contains:

```python
from nltk.tokenize.sonority_sequencing import punctuation
from nltk.lm import Vocabulary
from nltk.corpus import wordnet
from nltk.stem.wordnet import WordNetLemmatizer
from nltk import word_tokenize, pos_tag, sent_tokenize
from collections import defaultdict
from nltk.corpus import stopwords
from nltk.util import ngrams, bigrams
```

```python
# Open the file and read its contents.
with open('article2.txt', encoding='utf8') as file:
       text = file.read()
```

```python
from nltk.stem import WordNetLemmatizer

def get_wordnet_pos(word):
    """Map POS tag to first character lemmatize() accepts"""
    tag = nltk.pos_tag([word])[0][1][0].upper()
    tag_dict = {"J": wordnet.ADJ,
                "N": wordnet.NOUN,
                "V": wordnet.VERB,
                "R": wordnet.ADV}
    return tag_dict.get(tag, wordnet.NOUN)

text_list=word_tokenize(text) # Split the text into words.

#Remove stop words.
stop_words=stopwords.words('english')
text_list=[word for word in text_list if word not in stop_words]

# Find the lemmas of the words and combine them into a text.
lemma_function = WordNetLemmatizer()
text1 = ' '.join([lemma_function.lemmatize(w, get_wordnet_pos(w)) for w in text_list])

text1=text1.lower() # Convert all words to lowercase.

text_sent=sent_tokenize(text1) # Split the text into sentences.

for p in string.punctuation: # Remove punctuation from the sentences.
    for i in range(len(text_sent)):
      if p in text_sent[i]:
        text_sent[i] = text_sent[i].replace(p, '') # Replace a character in a string.

preprocessed_text=' '.join(text_sent) # Preprocessed text.

print(text_sent)
print(preprocessed_text)
```

```python
# Find the 5 most frequently used words in the text.
from nltk.probability import FreqDist

# Tokenize the preprocessed text into words.
word_preprocessed_text=word_tokenize(preprocessed_text)

fdist=FreqDist(word_preprocessed_text)
keywords=[word for word,freq in fdist.most_common(6)]
print(keywords)
```

```python
# Creates a sequence of words in descending order of their frequency,
# with the most frequent word appearing first in the text.
from nltk.probability import FreqDist

fdist=FreqDist(word_preprocessed_text)
for word in word_preprocessed_text:
    fdist[word] += 1

print(*fdist) # * operator allows the elements of fdist to be passed to the print() function as separate arguments.
```

```python
from nltk import bigrams, word_tokenize
from nltk.lm.preprocessing import flatten, pad_both_ends

bigrams = __import__('nltk').bigrams
#Эта строка восстанавливает функцию bigrams из библиотеки NLTK,
#если ранее её имя было случайно переопределено, например списком.


padded_bigrams=[]
unigram =[]

for i in range(len(text_sent)):
    padded_bigrams.append(list(bigrams(pad_both_ends(word_tokenize(text_sent[i]), n=2))))
    text_word.append(word_tokenize(text_sent[i]))

unigram=list(flatten(pad_both_ends(sent, n=2) for sent in text_word))

print("Биграммы: \n", padded_bigrams)
print("Униграммы: \n", unigram)
```

```python
# Create bigrams from the entire text.
bigrams = list(ngrams(word_preprocessed_text, 2))

# Count the number of occurring bigrams.
frequency = Counter(bigrams)

# Count the occurrences of each first word in the bigram.
first_word_count = Counter([bigram[0] for bigram in bigrams])

# Calculate the conditional probability for each bigram.
conditional_probabilities = {
    bigram: count/first_word_count[bigram[0]]
    for bigram, count in frequency.items()
}
```

```python
# Output the results.
for bigram, probability in conditional_probabilities.items():
    print(f"Bigram: {bigram}, P = {probability}")
```

```python
# Calculate the probability of sentences.
sentences_probability=[]

for i in range(len(text_sent)):
    sentence_bigrams = list(ngrams(word_tokenize(text_sent[i]), 2))
    sentence_probability = 1.0
    for bigram in sentence_bigrams:
        if bigram in conditional_probabilities:
            sentence_probability *= conditional_probabilities[bigram]
            sentences_probability.append(sentence_probability)

for i in range(len(sent_tokenize(text))):
    print(f' {sent_tokenize(text)[i]} = ', sentences_probability[i])
```

```python
the_dict = {text: probability for text, probability in zip(sent_tokenize(text), sentences_probability)}
the_dict = sorted(the_dict.items(), key=lambda pair: pair[1], reverse=True)
for i in the_dict:
  print(i)
```

```python
import nltk
nltk.download('punkt_tab')
```

```python
with open('article2.txt', encoding='utf8') as file:
       text = file.read()

# Load Packages
from sumy.parsers.plaintext import PlaintextParser
from sumy.nlp.tokenizers import Tokenizer
from sumy.summarizers.text_rank import TextRankSummarizer


# For Strings
parser = PlaintextParser.from_string(text,Tokenizer("english"))

# Summarize using sumy TextRank
summarizer = TextRankSummarizer()
summary =summarizer(parser.document,2)
text_summary=""
for sentence in summary:
    text_summary+=str(sentence)
```
