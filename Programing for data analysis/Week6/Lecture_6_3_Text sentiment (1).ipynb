# **TextBlob** is a user-friendly object from the Python library of the same name that is used for text processing. It supports a wide range of natural language processing (NLP) tasks, such as parsing, sentiment analysis, translation, and more.

# The .sentiment.polarity attribute of a TextBlob object is used to determine the polarity of a text, that is, its emotional tone (positive, negative, or neutral). It returns a value in the range from -1 to 1, where:
# -1 indicates the most negative sentiment (very negative text),
# 1 indicates the most positive sentiment (very positive text),
# 0 indicates neutral text.

# **TextBlob **— это удобный в использовании объект из одноименной библиотеки Python, который применяется для обработки текстов. Он поддерживает широкий спектр задач в области обработки естественного языка (NLP), таких как синтаксический разбор, анализ тональности, переводы и многое другое.

Атрибут .sentiment.polarity объекта TextBlob используется для определения полярности текста, то есть его эмоциональной окраски (положительной, отрицательной или нейтральной). Он возвращает значение в диапазоне от -1 до 1, где:

-1 — означает максимально отрицательную тональность (очень негативный текст),

1 — максимально положительную тональность (очень позитивный текст),

0 — нейтральный текст.

```python
from textblob import TextBlob

def determine_sentiment(text):
    # Create a TextBlob object with the analyzed text.
    text_analysis = TextBlob(text)

    # Determine the sentiment.
    if text_analysis.sentiment.polarity > 0.01:
        return "Positive"
    elif text_analysis.sentiment.polarity < -0.01:
        return "Negative"
    else:
        return "Neutral"


sentence = input("Enter a sentence for analysis: ")
sentiment = determine_sentiment(sentence)

print(f"Sentence sentiment: {sentiment}")
```

Библиотека TextBlob предназначена преимущественно для обработки текстов на английском языке. Она предоставляет инструменты токенизации, определения частей речи, лемматизации, анализа тональности и других задач NLP. Однако стандартные модели TextBlob не обеспечивают полноценную лингвистическую обработку русского языка. Для работы с русскоязычными текстами рекомендуется использовать специализированные инструменты, например spaCy с моделями русского языка (ru_core_news_sm, ru_core_news_md, ru_core_news_lg). Альтернативой может служить библиотека Polyglot, поддерживающая многоязычную обработку текста, включая русский язык, при условии установки соответствующих моделей.

# **VaderSentiment** is a text sentiment analysis library that uses a model called VADER (Valence Aware Dictionary and sEntiment Reasoner).
# This model is specifically designed for sentiment analysis of texts, especially in the context of social media and short messages (e.g., tweets).

# VADER takes into account both word polarity and the intensity of emotions expressed through emojis, capitalization, and punctuation (e.g., exclamation marks).

# Features of VADER:

# Works well with texts containing informal expressions, slang, and emojis.

# Processes large volumes of data quickly, as it is based on precompiled lexicons.

# Considers contextual intensifiers or negations (e.g., words like “very”, “not”, etc.).

# **vaderSentiment** — это библиотека для анализа тональности текста, которая использует модель под названием VADER (Valence Aware Dictionary and sEntiment Reasoner).
# Эта модель специально разработана для анализа тональности текстов, особенно в контексте социальных сетей и коротких сообщений (например, твитов).
# VADER учитывает как полярность слов, так и интенсивность эмоций, обозначенных эмодзи, заглавными буквами и пунктуацией (например, восклицательные знаки).

# Особенности VADER:
# Хорошо работает с текстами, содержащими разговорные выражения, сленг и эмодзи.
# Быстро обрабатывает большие объемы данных, так как основывается на заранее составленных словарях.
# Учитывает контекст усиления или ослабления тональности (например, слова "очень", "не" и т.п.).

```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

# Load the pre-trained spaCy model for the English language.
nlp = spacy.load("en_core_web_sm")

# Create a sentiment analyzer using vaderSentiment.
analyzer = SentimentIntensityAnalyzer()

def determine_sentiment(text):
    # Process the text using spaCy.
    doc = nlp(text)
    sentiment = analyzer.polarity_scores(text)# Calculate sentiment using vaderSentiment.

    # Determine the sentiment based on the analysis results.
    if sentiment['compound'] >= 0.05:
        return "Positive"
    elif sentiment['compound'] <= -0.05:
        return "Negative"
    else:
        return "Neutral"

sentence = input("Enter a sentence for analysis: ")
sentiment = determine_sentiment(sentence)
print(f"Sentence sentiment: {sentiment}")
```

# **§3. Text sentiment analysis using the NLTK library. / Анализ тональности текста с использованием библиотеки nltk**

# The output will contain values for the following keys:
# neg: the proportion of negative sentiment in the text,
# neu: the proportion of neutral sentiment,
# pos: the proportion of positive sentiment,
# compound: the overall sentiment score, representing an aggregated value that ranges from -1 (most negative text) to 1 (most positive text).

# Вывод будет содержать значения для следующих ключей:

# neg: доля негативных эмоций в тексте,
# neu: доля нейтральных эмоций,
# pos: доля позитивных эмоций,
# compound: итоговая оценка тональности, представляющая собой агрегированное значение, которое может варьироваться от -1 (максимально негативный текст) до 1 (максимально позитивный текст).

```python
import nltk
from nltk.sentiment.vader import SentimentIntensityAnalyzer

# Загрузите пакет данных для VADER Sentiment Analyzer
nltk.download('vader_lexicon')

# Create a sentiment analyzer.
analyzer = SentimentIntensityAnalyzer()

def determine_sentiment(text):
    # Analyze the text and obtain sentiment scores.
    sentiment_scores = analyzer.polarity_scores(text)
    print(sentiment_scores)
    # Determine the sentiment based on the compound score.
    compound_score = sentiment_scores['compound']
    if compound_score >= 0.05:
        return "Positive"
    elif compound_score <= -0.05:
        return "Negative"
    else:
        return "Neutral"

sentence = input("Enter a sentence for analysis: ")
sentiment = determine_sentiment(sentence)
print(f"Sentence sentiment: {sentiment}")
```

# **§4. Text sentiment analysis using TfidfVectorizer from the scikit-learn (sklearn) library./ Анализ тональности текста с использованием TfidfVectorizer из библиотеки scikit-learn (sklearn)**
# TfidfVectorizer from the scikit-learn (sklearn) library is used to create vector representations of texts based on the TF–IDF (Term Frequency–Inverse Document Frequency) statistic. However, it does not provide built-in tools for sentiment detection by itself. To perform sentiment analysis using TfidfVectorizer, you also need to train a machine learning model, such as a Support Vector Machine (SVM) classifier or a Naive Bayes classifier.
# Below is an example Python program that uses TfidfVectorizer and SVM to determine sentiment in English text. Please note that to train the model you will need a labeled sentiment dataset.
# TfidfVectorizer из библиотеки scikit-learn (sklearn) используется для создания векторных представлений текстов на основе статистической меры TF-IDF (Term Frequency-Inverse Document Frequency). Однако он не предоставляет непосредственных инструментов для определения тональности текста. Для анализа тональности текста с использованием TfidfVectorizer, вам также потребуется обучить модель машинного обучения, такую как классификатор на основе метода опорных векторов (SVM) или наивного байесовского классификатора.

# Вот пример программы на Python, которая использует TfidfVectorizer и SVM для определения тональности текста на английском языке. Пожалуйста, учтите, что для обучения модели вам потребуется набор данных с размеченными данными по тональности.



```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn import svm
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report
```

```python
# Labeled data: 30 sentences with sentiment labels (1 – positive, 0 – negative).
data = [
    ("I love this product, it's amazing!", 1),
    ("This is the best experience I've ever had!", 1),
    ("I'm very satisfied with the service.", 1),
    ("Everything was perfect, highly recommend.", 1),
    ("The food was delicious and the staff was friendly.", 1),
    ("Great atmosphere and excellent service.", 1),
    ("This was a fantastic movie!", 1),
    ("The book is absolutely incredible!", 1),
    ("I really enjoyed the vacation, it was relaxing and fun.", 1),
    ("The presentation was excellent and very informative.", 1),
    ("I had an awful experience.", 0),
    ("This is the worst product I've ever bought.", 0),
    ("I am extremely disappointed.", 0),
    ("The service was terrible and the staff was rude.", 0),
    ("The quality of the product is really poor.", 0),
    ("I will never come back to this place again.", 0),
    ("The food was cold and tasteless.", 0),
    ("I hated the experience, it was a waste of money.", 0),
    ("It was a boring and pointless movie.", 0),
    ("The product broke after one day of use.", 0),
    ("The website is too slow and difficult to navigate.", 0),
    ("The room was dirty and uncomfortable.", 0),
    ("The customer service was unresponsive and unhelpful.", 0),
    ("The packaging was damaged and the product didn't work.", 0),
    ("I am very unhappy with this purchase.", 0),
    ("The app is very user-friendly and intuitive.", 1),
    ("I am absolutely thrilled with the results!", 1),
    ("The staff went above and beyond to help.", 1),
    ("The product exceeded all my expectations.", 1),
    ("The design is elegant and the functionality is great.", 1)
]

# Split the data into texts and labels.
texts = [item[0] for item in data]
labels = [item[1] for item in data]

# Convert texts into TF–IDF vectors.
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(texts)

# Split the data into training and test sets.
X_train, X_test, y_train, y_test = train_test_split(X, labels, test_size=0.2, random_state=42)

# Create and train an SVM model.
model = svm.SVC(kernel='linear')
model.fit(X_train, y_train)

# Prediction on test data
y_pred = model.predict(X_test)

# Model performance evaluation
accuracy = accuracy_score(y_test, y_pred)
print(f"Model accuracy: {accuracy * 100:.2f}%")

# Detailed classification metrics report
print("\nClassification report:")
print(classification_report(y_test, y_pred))

# Example of prediction on a new text
new_text = ["The product exceeded all my expectations"]
new_text_vectorized = vectorizer.transform(new_text)
prediction = model.predict(new_text_vectorized)

# Prediction result
print("\nSentiment of the new text:", "Positive" if prediction[0] == 1 else "Negative")
print()
```

Как работает программа:
Данные: Мы создаём размеченный набор данных из 20 предложений с тональностью (положительная — 1, отрицательная — 0).
TfidfVectorizer: Преобразует текстовые данные в векторы признаков, используя TF-IDF (Term Frequency-Inverse Document Frequency).
SVM: Модель опорных векторов (SVM) используется для классификации данных на основе признаков.
Разделение данных: Мы делим набор данных на обучающую (70%) и тестовую (30%) выборки.
Обучение и оценка: Модель обучается на обучающих данных, затем проверяется на тестовой выборке. Точность модели и отчёт по классификации выводятся в консоль.
Предсказание: Программа может также делать предсказания на новых текстах.

Параметр random_state=42 в функции train_test_split (и в других функциях, связанных со случайными процессами в библиотеке scikit-learn) используется для того, чтобы зафиксировать генерацию случайных чисел и обеспечить повторяемость эксперимента.

Многие алгоритмы машинного обучения, такие как разделение данных на обучающую и тестовую выборки, или сам процесс обучения моделей (например, в случайном лесу), используют случайные числа. Чтобы результаты не менялись при каждом запуске программы, и для воспроизводимости экспериментов, мы устанавливаем значение random_state.

random_state задает начальное состояние генератора случайных чисел. Это как "посев" (seed) для генерации случайных чисел.

Если установить random_state=42, каждый раз, когда вы запускаете код, случайные числа будут генерироваться одинаково, что приведет к одинаковому разделению данных или одинаковому результату.

Число 42 стало неофициальным стандартом для значения random_state в обучении машинного обучения. Это отсылка к книге Дугласа Адамса "Автостопом по галактике", где 42 — это "ответ на главный вопрос жизни, Вселенной и всего такого". Однако это всего лишь шутка, и вы можете использовать любое другое значение для random_state.
