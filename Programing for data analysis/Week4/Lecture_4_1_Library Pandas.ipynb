```python
# Замена средним значением / Filling with the Mean
mean_english = df_clean["english_score"].mean()

df_clean["english_score"] = df_clean["english_score"].fillna(mean_english)
```

# Missing scores are replaced with the mean of the available scores.
# Пропущенные оценки заменяются средним значением существующих оценок.

```python
# Mean or Median? / Среднее или медиана?
df_clean["programming_score"] = df_clean["programming_score"].fillna(
    df_clean["programming_score"].median()
)
```
