# These commands can produce similar results, but they use different selection mechanisms.

```python
# Фильтрация по условию / Filtering by a Condition
df_clean["final_score"] > 85
```

# Pandas evaluates the condition for each row and returns True or False.

```python
# Boolean Mask / Логическая маска
high_score = df_clean["final_score"] > 85
```

```python
df_clean[high_score][["student_id", "final_score"]]
```

```python
df_clean.loc[high_score, ["student_id", "final_score"]]
```

# A Boolean mask is a sequence of True and False values that determines which rows should be selected.

```python
# Фильтрация категориальных данных / Filtering Categorical Data
df_clean[df_clean["program"] == "Data Science"]
```

```python
df_clean[df_clean["city"] == "Astana"]
```

```python
# Оператор != / Not Equal Operator
df_clean[df_clean["program"] != "Data Science"]
```

```python
# Несколько условий: AND / Multiple Conditions: AND
df_clean[
    (df_clean["program"] == "Data Science") &
    (df_clean["final_score"] > 85)
]
```

```python
# Несколько условий: OR / Multiple Conditions: OR
df_clean[
    (df_clean["program"] == "Data Science") |
    (df_clean["program"] == "Cybersecurity")
]
```

```python
# Отрицание условия / Negating a Condition
df_clean[
    ~(df_clean["program"] == "Data Science")
]
```

# ~ инвертирует Boolean mask: True становится False, а False становится True.
# ~ reverses a Boolean mask: True becomes False, and False becomes True.

```python
# Combining Three Conditions
df_clean[
    (df_clean["program"] == "Data Science") &
    (df_clean["attendance"] >= 80) &
    (df_clean["final_score"] > 85)
]
```

```python
# Filtering and Selecting Specific Columns
# Фильтрация и выбор нужных столбцов
df_clean.loc[
    df_clean["final_score"] > 85,
    ["student_id", "program", "attendance", "final_score"]
]
```

```python
# isin() — проверка принадлежности множеству
# isin() — Membership Testing
df_clean[
    (df_clean["program"] == "Data Science") |
    (df_clean["program"] == "Cybersecurity")
]
```

```python
# significantly simpler
df_clean[
    df_clean["program"].isin(
        ["Data Science", "Cybersecurity"]
    )
]
```

# isin() checks whether a value belongs to one of the elements in the specified list.
# isin() проверяет, принадлежит ли значение одному из элементов указанного списка.

```python
# ~isin() — исключение категорий / ~isin() — Excluding Categories
df_clean[
    ~df_clean["program"].isin(
        ["Data Science", "Cybersecurity"]
    )
]
```

# The ~ operator reverses the result of isin().

```python
# between() — фильтрация диапазона / between() — Range Filtering
df_clean[
    (df_clean["final_score"] >= 80) &
    (df_clean["final_score"] <= 90)
]
```

```python
df_clean[
    df_clean["final_score"].between(80, 90)
]
```

```python
# Комбинация isin() и between() / Combining isin() and between()
# Find students in Data Science or Software Engineering whose
# final score is between 80 and 90.
df_clean[
    df_clean["program"].isin(
        ["Data Science", "Software Engineering"]
    )
    &
    df_clean["final_score"].between(80, 90)
]
```

```python
# Выбор одной конкретной ячейки / Selecting a Single Value
df_clean.loc[0, "final_score"]
```

```python
df_clean.iloc[0, 11]
```

# The first example uses labels, while the second uses positions.

```python
# Сортировка результата после фильтрации /
# Sorting Filtered Results
df_clean[
    df_clean["final_score"] >= 85
].sort_values("final_score", ascending=False)
```

# First, we select students with a final score of at least 85, and then sort them from the highest to the lowest score.
# Сначала выбираем студентов с итоговым баллом не менее 85, затем сортируем их от максимального балла к минимальному.

```python
# Programs are sorted in ascending order,
# while scores within each program are sorted in descending order.
df_clean.sort_values(
    ["program", "final_score"],
    ascending=[True, False]
)
```

# **Создадим категорию академической успеваемости студента.**
# **Let's create an academic performance category based on the student's final score.**

```python
# Добавление категориального столбца / Adding a Categorical Column
def performance_level(score):
    if score >= 90:
        return "Excellent"
    elif score >= 75:
        return "Good"
    elif score >= 60:
        return "Satisfactory"
    else:
        return "Low"

df_clean["performance_level"] = df_clean["final_score"].apply(performance_level)
```

# LabelEncoder assigns a numerical label to each category.

```python
# OneHotEncoder
df_clean[["program"]].head()

onehot_encoder = OneHotEncoder(
    sparse_output=False
)

program_encoded = onehot_encoder.fit_transform(
    df_clean[["program"]]
)
onehot_encoder.get_feature_names_out(["program"])
```

# sparse_output=False означает: вернуть обычный  NumPy-массив (dense array).В One-Hot Encoding обычно очень много нулей. Поэтому вместо хранения каждого 0 sparse matrix хранит данные более экономно.
# False → обычный массив, удобнее для демонстрации.
# True → разреженная матрица, эффективнее для больших данных.
# sparse_output=False means: return a regular  NumPy array.One-hot encoded data usually contain many zeros. A sparse matrix stores such data more efficiently instead of explicitly storing every zero.

```python
program_encoded_df = pd.DataFrame(
    program_encoded,
    columns=onehot_encoder.get_feature_names_out(["program"]),
    index=df_clean.index
)

program_encoded_df.head()
```
