# **§2.Creation and Initialization of Arrays / Создание и заполнение массивов**


```python
import numpy as np
import matplotlib.pyplot as plt
```

# Create a ratings matrix:
# R = np.zeros((20, 12), dtype=np.int8)

```python
# Part 1. Create & fill arrays
# ---------------------------
n_users, n_items = 20, 12

R = np.zeros((n_users, n_items), dtype=np.int8)  # ratings matrix
print(R)
```

# Create an array of course names (strings) and a user_id array:
# courses = np.array([...12 course names...])
# users = np.arange(1, 21)

```python
users = np.arange(1, n_users + 1)
courses = np.array([
    "Python Basics", "Data Analysis", "SQL", "ML Intro", "Deep Learning", "NLP",
    "Computer Vision", "Cloud", "Cybersecurity", "Project Mgmt", "DevOps", "GIS"
], dtype=object)

print("R shape:", R.shape, "dtype:", R.dtype)
print(users)
print(courses)
```

# Fill the data using different methods:
# randomly select 200 (user, course) pairs and assign a random rating from 1 to 5
# hint: np.random.default_rng(seed) + integers + indexing



```python
# Fill with random interactions: choose 200 (user, item) pairs and set rating 1..5
rng = np.random.default_rng(42)
m = 200
u_idx = rng.integers(0, n_users, size=m) # rng.integers(0, n_users, size=m) generates m random user indices from 0 to n_users - 1.
i_idx = rng.integers(0, n_items, size=m)
ratings = rng.integers(1, 6, size=m)  # 1..5

# If same pair repeats, last one overwrites (OK for demo)
R[u_idx, i_idx] = ratings

print(R)
```

# alternatively: manually fill the first 5 users using slicing

```python
print(R[:5, :3])

# Also manual fill via slicing for first 5 users (just to show)
R[:5, :3] = rng.integers(1, 6, size=(5, 3))
print(R[:5, :3])
```

# Save and load the array:
# np.save("ratings.npy", R)
# R2 = np.load("ratings.npy")
# Check:np.array_equal(R, R2)

```python
# Save/load
np.save("ratings.npy", R)
R2 = np.load("ratings.npy")
print("Saved & loaded OK:", np.array_equal(R, R2))
print()
print(R2)
```

# *Display*:

```python
# Part 2. Indexing & editing
# ---------------------------
u, c = 10, 3           # human-readable (1-based)
ui, ci = u - 1, c - 1  # numpy indices (0-based)

print("\nRating user 10 for course 3:", R[ui, ci])
print("All ratings of user 10:", R[ui, :])
print("All ratings for course 3:", R[:, ci])
```

# *Modifications:*
# • user 10 completed course 3 and gave a rating of 5 → update the element
# • user 7 removed the rating for course 2 → set it back to 0
# • increase the rating in course 1 by 1 for all users, but not above 5
# (hint: np.clip)

```python
# Updates
print(R[ui, ci])
R[ui, ci] = 5         # user 10 sets 5
print(R[ui, ci])

print(R[7 - 1, 2 - 1])
R[7 - 1, 2 - 1] = 0   # user 7 removes rating for course 2
print(R[7 - 1, 2 - 1])
```

```python
# Increase all ratings in course 1 by 1, but not above 5 (keep 0 as 0)
col0 = R[:, 0]
print(R[:, 0])
mask_taken_col0 = col0 > 0
print(mask_taken_col0)

#The values ​​must remain no less than 1 and no more than 5.
col0[mask_taken_col0] = np.clip(col0[mask_taken_col0] + 1, 1, 5)
R[:, 0] = col0

print(R[:, 0])
```

# *Index search:*
# • find the indices of all non-zero ratings of user 10
# (np.nonzero, np.where)

```python
# Find indices of nonzero ratings for user 10
nz_items_user10 = np.nonzero(R[ui, :])[0] # We take the first element of the tuple.
print("\nUser 10 took course indices (0-based):", nz_items_user10)
print("User 10 took courses:", courses[nz_items_user10])
```

```python
# Part 3. Masks, filtering, counts
# ---------------------------
taken = R > 0

courses_per_user = taken.sum(axis=1)
users_per_course = taken.sum(axis=0)

print("\nCourses taken per user (first 5):", courses_per_user[:5])
print("Users per course:", users_per_course)
```

# *Calculate:*
# • how many courses each user has completed → taken.sum(axis=1)
# • how many users completed each course → taken.sum(axis=0)
# • the share of “high” ratings (>=4) for each course among those who completed it
# o carefully divide by the number of completed entries (protection from division by zero)

image.png

```python
# Share of high ratings per course among taken

high = R >= 4

high_count = (high & taken).sum(axis=0)
taken_count = taken.sum(axis=0)
high_share = np.divide(high_count, taken_count, # avoid division by zero
                       out=np.zeros_like(high_count, dtype=float), # creates an array filled with zeros
                       where=taken_count > 0) # mask

print("High-rating share per course:", np.round(high_share, 3))
```

# *Filtering*:
# • output the list of courses where the average rating (among completed users) is >= 3
# • find users who have completed ≥ 8 courses

```python
# Filter: courses with mean rating >= 3 among taken
# (use NaN method)
R_nan = R.astype(np.float32).copy()
R_nan[R_nan == 0] = np.nan
mean_course = np.nanmean(R_nan, axis=0) # это функция NumPy для вычисления среднего значения с игнорированием NaN (Not a Number).

good_courses = courses[mean_course >= 3]
print("\nCourses with mean >= 3:", good_courses)

# Users who took >= 8 courses
active_users_idx = np.where(courses_per_user >= 8)[0]
print("Users with >=8 courses:", users[active_users_idx])
```

# **Average ratings:**
# • average rating per user (only non-zero values)
# • average rating per course (only non-zero values)

# **Hints:**
# • option A: replace 0 with np.nan and compute using np.nanmean
# • option B: compute the sum and divide by taken.sum, where taken.sum > 0

```python
# Part 4. Aggregations excluding zeros (2 approaches)
# ---------------------------
# A) NaN-based (already have R_nan)
mean_user_nan = np.nanmean(R_nan, axis=1)
mean_course_nan = np.nanmean(R_nan, axis=0)

# B) sum/count with mask
sum_user = R.sum(axis=1)
cnt_user = taken.sum(axis=1)
mean_user_mask = np.divide(sum_user, cnt_user,
                           out=np.zeros(n_users, dtype=float),
                           where=cnt_user > 0)

sum_course = R.sum(axis=0)
cnt_course = taken.sum(axis=0)
mean_course_mask = np.divide(sum_course, cnt_course,
                             out=np.zeros(n_items, dtype=float),
                             where=cnt_course > 0)

print("\nMean rating by user (first 5) NaN:", np.round(mean_user_nan[:5], 2))
print("Mean rating by user (first 5) mask:", np.round(mean_user_mask[:5], 2))
```

```python
# Top-3 courses by mean rating
top3_idx = np.argsort(mean_course_mask)[-3:][::-1]
print("\nTop-3 courses by mean:",
      list(zip(courses[top3_idx],
               np.round(mean_course_mask[top3_idx], 2))))

# Most taken course
most_taken_idx = np.argmax(users_per_course)
print("Most taken course:", courses[most_taken_idx],
      "taken by", users_per_course[most_taken_idx], "users")
```

# **§5. Normalization and Transformations (Important for ML)/ Нормализация и преобразования (важно для ML)**



# **Нормализация** — это преобразование данных к стандартизированной шкале, чтобы убрать систематические смещения, сделать значения сопоставимыми и улучшить качество анализа, обучения моделей и сравнения объектов.

# *Что это означает по сути*

# Каждый пользователь имеет свою шкалу оценивания:
# один всегда ставит высокие оценки (4–5),
# другой — более строгий (2–3),
# третий — использует весь диапазон.
# 👉 Модель начинает путать вкусы с личным стилем оценивания.
# Нормализация устраняет этот эффект.

image.png

# *После нормализации:*

# 0 → нейтральное отношение
# > 0 → нравится больше среднего
# < 0 → нравится меньше среднего
# То есть оценки становятся относительными, а не абсолютными.

# **ПРИМЕР**
# User A: [5, 4, 5]
# mean = 4.67
# → centered = [0.33, -0.67, 0.33]

Задание: Нормализация оценок по пользователям (centered ratings):
•	R_centered[u, i] = R[u, i] - mean_u, но только там, где R>0
Стандартизация по курсам (z-score) для ненулевых:
•	использовать среднее и стандартное отклонение курса
•	защититься от std=0
Проверка:
•	среднее по пользователю после центрирования должно быть около 0 (по ненулевым)

# User-based normalization of ratings (centered ratings):
# • R_centered[u, i] = R[u, i] - mean_u, but only where R > 0

```python
# Part 5. Normalization
# ---------------------------
# 5.1 Center ratings per user (subtract user's mean from their nonzero ratings)
mean_u = mean_user_mask  # per user mean excluding zeros
mean_u = mean_user_mask.astype(np.float32)      # (n_users,)
R_centered = R.astype(np.float32) - mean_u[:, None]  # (n_users, n_items) - (n_users,1)
R_centered[~taken] = 0.0

# Check: mean per user on nonzeros ~0
sum_center = R_centered.sum(axis=1)
cnt = taken.sum(axis=1)
mean_center = np.divide(sum_center, cnt, out=np.zeros(n_users, dtype=float), where=cnt > 0)
print("\nCentered mean per user (first 10):", np.round(mean_center[:10], 4))
```

# **Стандартизация** по курсам (z-score normalization) — это метод приведения оценок/признаков к нулевому среднему и единичному стандартному отклонению для каждого курса (item / feature), чтобы сделать данные сопоставимыми по шкале и убрать масштабные и дисперсионные искажения.

# Стандартизация по курсам (z-score) =
# преобразование оценок каждого курса в относительные значения (σ-единицы),
# устраняет масштаб и дисперсию,
# делает курсы статистически сопоставимыми,
# критически важна для similarity-based моделей.

# Course-based standardization (z-score) for non-zero values:
# • use the course mean and standard deviation
# • protect against std = 0

# Validation:
# • the mean per user after centering should be approximately 0 (for non-zero values only)

```python
# R: (n_users, n_items) ratings / interactions matrix (0 = missing)
# taken: (n_users, n_items) boolean mask, True where R > 0
# n_users, n_items inferred from R

R_float = R.astype(np.float32)
n_users, n_items = R_float.shape

# 1) Count nonzero ratings per course (item)
cnt_course = taken.sum(axis=0).astype(np.int32)  # (n_items,)

# 2) Mean per course excluding zeros
sum_i = np.where(taken, R_float, 0.0).sum(axis=0)  # (n_items,)
mean_i = np.divide(
    sum_i, cnt_course,
    out=np.zeros(n_items, dtype=np.float32),
    where=cnt_course > 0
)  # (n_items,)

# 3) Variance per course excluding zeros (population variance; for sample use (cnt_course-1))
diff2 = (R_float - mean_i[None, :]) ** 2         # (n_users, n_items)
diff2[~taken] = 0.0                              # ignore missing

var_i = np.divide(
    diff2.sum(axis=0), cnt_course,
    out=np.zeros(n_items, dtype=np.float32),
    where=cnt_course > 0
)  # (n_items,)

std_i = np.sqrt(var_i).astype(np.float32)

# 4) Safe std to avoid division by zero
std_safe = np.where(std_i < 1e-8, 1.0, std_i).astype(np.float32)  # (n_items,)

# 5) Z-score standardization per course (item-wise), then keep missing as 0
R_z = (R_float - mean_i[None, :]) / std_safe[None, :]  # broadcasting works
R_z[~taken] = 0.0

# 6) Optional check: mean ~0 and std ~1 per course on observed entries
# Mean check
mean_check = np.divide(
    np.where(taken, R_z, 0.0).sum(axis=0), cnt_course,
    out=np.zeros(n_items, dtype=np.float32),
    where=cnt_course > 0
)

# Std check
diff2_check = (R_z - mean_check[None, :]) ** 2
diff2_check[~taken] = 0.0
var_check = np.divide(
    diff2_check.sum(axis=0), cnt_course,
    out=np.zeros(n_items, dtype=np.float32),
    where=cnt_course > 0
)
std_check = np.sqrt(var_check)

print("Z-score mean per course (first 10):", np.round(mean_check[:10], 4))
print("Z-score std  per course (first 10):", np.round(std_check[:10], 4))
```
