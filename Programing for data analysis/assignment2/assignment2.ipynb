### 2.5. Сравнение городов и интерпретация — 4 балла

1. Преобразуйте данные каждого города в последовательность из 168 температур (7 × 24).
2. Постройте корреляционную матрицу 3 × 3 с помощью `np.corrcoef`.
3. Не учитывая диагональ, определите пару городов с наибольшим коэффициентом корреляции.
4. Напишите 2–3 предложения, что означает положительная, отрицательная и близкая к нулю корреляция в контексте температур.

```python
T_city = T.reshape(168, 3).T
corr_maytix = np.corrcoef(T_city)
rows, cols = np.triu_indices(3, k=1)
print(rows, cols)
most_corr_idx = corr_maytix[rows, cols].argmax()
print(most_corr_idx)
city_1 = rows[most_corr_idx]
city_2 = cols[most_corr_idx]
print('Cities:', city_1 + 1, city_2 + 1)
print('Correlation:', corr_maytix[city_1, city_2])
```

Положительная корреляция означает, что температуры двух городов обычно повышаются и понижаются вместе. Отрицательная — что температура в одном городе повышается, а в другом обычно понижается. Близкая к нулю корреляция означает, что нет выраженной линейной связи: повышение или понижение температуры в одном городе не сопровождается определённым изменением в другом.
