Близкая к нулю корреляция означает, что нет выраженной линейной связи: повышение или понижение температуры в одном городе не сопровождается определённым изменением в другом.

```python
print('1. Корреляция признаков с ценой')

price_correlation = df.select_dtypes(include='number').corr()['price']
price_correlation = price_correlation.drop('price').dropna()
price_correlation = price_correlation.loc[
    price_correlation.abs().sort_values(ascending=False).index
]

display(price_correlation.round(3))
```

```python
print('2. Положительная и отрицательная корреляция')

positive = price_correlation[price_correlation > 0]
negative = price_correlation[price_correlation < 0]

if not positive.empty:
    feature = positive.idxmax()
    print(f'Положительная: {feature}, коэффициент {positive[feature]:.3f}')
else:
    print('Положительной корреляции нет.')

if not negative.empty:
    feature = negative.idxmin()
    print(f'Отрицательная: {feature}, коэффициент {negative[feature]:.3f}')
else:
    print('Отрицательной корреляции нет.')
```
