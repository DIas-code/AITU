3. Создайте `rooms_array` — 200 целых значений количества комнат от 1 до 5 включительно.
4. Создайте `price_array` — 200 значений стоимости квартир в диапазоне от 15 до 120 млн тенге.

> **Подсказка:** Для генерации можно использовать `np.random.randint()`, `np.random.uniform()` или методы объекта `rng`. Значения `price_array` допускается округлить до 1–2 знаков.

**Что должно быть в результате:**
- Три массива длиной 200.

```python
import numpy as np
```

```python
rng = np.random.default_rng(42)
n = 200
area_array = np.round(rng.uniform(25, 150, size=n),2)
rooms_array = rng.integers(1, 6, size=n)
price_array = np.round(rng.uniform(15, 120, size=n),2)
print(len(area_array), len(rooms_array), len(price_array))
```

```python
print(f'area_array: shape: {area_array.shape}, ndim: {area_array.ndim} size: {area_array.size} dtype: {area_array.dtype}')
print(f'rooms_array: shape: {rooms_array.shape}, ndim: {rooms_array.ndim} size: {rooms_array.size} dtype: {rooms_array.dtype}')
print(f'price_array: shape: {price_array.shape}, ndim: {price_array.ndim} size: {price_array.size} dtype: {price_array.dtype}')
```

shape - показывает количество элементов в dimensions
ndim - показвает количество dimensions 
size - общее количество элементов
dtype - тип данных в array

### 1.5. Нормализация и краткий вывод — 4 балла

1. Нормализуйте `area_array` и `price_array` методом Min-Max в диапазон [0, 1]. Формулу реализуйте самостоятельно средствами NumPy.
2. Проверьте минимальное и максимальное значения нормализованных массивов.
3. Выведите первые 5 исходных и первые 5 нормализованных значений.
4. Напишите 2–3 предложения: зачем нужна нормализация и в каких задачах она может быть полезна.

```python
area_normalized = (area_array - area_array.min()) / (area_array.max() - area_array.min())
price_normalized = (price_array - price_array.min()) / (price_array.max() - price_array.min())

print(f'area normalized min: {area_normalized.min():.2f}, max: {area_normalized.max():.2f}')
print(f'price normalized min: {price_normalized.min():.2f}, max: {price_normalized.max():.2f}')

print("\narea original:", area_array[:5])
print("area normalized:", area_normalized[:5])

print("\nprice original:", price_array[:5])
print("price normalized:", price_normalized[:5])
```

Нормализация приводит площадь и цену к общему диапазону от 0 до 1, сохраняя порядок значений. Она полезна в машинном обучении, например при кластеризации и поиске ближайших соседей, чтобы признаки с большими числовыми значениями не доминировали при расчёте расстояний.
