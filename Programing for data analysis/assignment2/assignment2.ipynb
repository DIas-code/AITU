### 1.1. Подготовка и создание данных — 4 балла

1. Импортируйте NumPy и задайте фиксированный seed для генератора случайных чисел.
2. Создайте `area_array` — 200 значений площади квартир в диапазоне от 25 до 150 м². Значения могут быть целыми или вещественными.
3. Создайте `rooms_array` — 200 целых значений количества комнат от 1 до 5 включительно.
4. Создайте `price_array` — 200 значений стоимости квартир в диапазоне от 15 до 120 млн тенге.
5. Убедитесь, что все три массива имеют одинаковое количество элементов и соответствуют одним и тем же 200 квартирам.
6. Для каждого массива выведите `shape`, `ndim`, `size` и `dtype`.

> **Подсказка:** Для генерации можно использовать `np.random.randint()`, `np.random.uniform()` или методы объекта `rng`. Значения `price_array` допускается округлить до 1–2 знаков.

**Что должно быть в результате:**
- Три массива длиной 200.
- Вывод основных атрибутов каждого массива.
- Короткое пояснение, что означают `shape`, `ndim`, `size` и `dtype`.

```python
import numpy as np
```

```python
rng = np.random.default_rng(42)
n = 200
area_array = np.round(rng.uniform(25, 150, size=n),2)
rooms_array = rng.integers(1, 6, size=n)
price_array = np.round(rng.uniform(15, 120, size=n),2)
print(len(area_array), len(rooms_array), len(price_array))
```

```python
print(f'area_array: shape: {area_array.shape}, ndim: {area_array.ndim} size: {area_array.size} dtype: {area_array.dtype}')
print(f'rooms_array: shape: {rooms_array.shape}, ndim: {rooms_array.ndim} size: {rooms_array.size} dtype: {rooms_array.dtype}')
print(f'price_array: shape: {price_array.shape}, ndim: {price_array.ndim} size: {price_array.size} dtype: {price_array.dtype}')
```

shape - показывает количество элементов в dimensions
ndim - показвает количество dimensions 
size - общее количество элементов
dtype - тип данных в array
