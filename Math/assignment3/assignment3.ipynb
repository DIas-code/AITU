### Part I. Mann–Whitney, U - Test

```python
def mann_whitney(group1, group2, U_crit):
    print("H0: распределения показателей в двух группах одинаковы.")
    print("H1: распределения показателей в двух группах различаются.")

    n1 = len(group1)
    n2 = len(group2)

    # Объединяем группы и сортируем
    combined = sorted(group1 + group2)

    # Ранги первой группы
    ranks1 = []
    for value in group1:
        rank = combined.index(value) + (combined.count(value) + 1) / 2
        ranks1.append(rank)

    # Ранги второй группы
    ranks2 = []
    for value in group2:
        rank = combined.index(value) + (combined.count(value) + 1) / 2
        ranks2.append(rank)

    # Суммы рангов
    R1 = sum(ranks1)
    R2 = sum(ranks2)

    # Вычисляем U по твоей формуле
    U1 = n1 * n2 + n1 * (n1 + 1) / 2 - R1
    U2 = n1 * n2 + n2 * (n2 + 1) / 2 - R2
    U_emp = min(U1, U2)

    print("\nn1 =", n1)
    print("n2 =", n2)
    print("Ранги первой группы:", ranks1)
    print("Ранги второй группы:", ranks2)
    print("R1 =", R1)
    print("R2 =", R2)
    print("U1 =", U1)
    print("U2 =", U2)
    print("U эмпирическое =", U_emp)
    print("U критическое =", U_crit)

    # Сравниваем с критическим значением
    if U_emp <= U_crit:
        print("\nU эмп ≤ U крит.")
        print("H0 отвергаем. H1 не отвергаем.")
        print("Различия статистически значимы.")
    else:
        print("\nU эмп > U крит.")
        print("H0 не отвергаем. H1 не подтверждается.")
        print("Статистически значимых различий не обнаружено.")

    return U_emp
```

```python
# Task 1. Comparing Student Performance Under Two Teaching Methods.
interactive = [78, 85, 90, 74, 88, 92]
traditional = [65, 72, 68, 60, 75, 70]

U_emp = mann_whitney(interactive, traditional, U_crit=5)
```

### Part II. Wilxocon test.

```python
def wilcoxon(before, after, T_crit):
    print("H0: систематических изменений нет.")
    print("H1: систематические изменения есть.")

    if len(before) != len(after):
        raise ValueError("Количество значений до и после должно совпадать.")

    # Разности: после минус до, без нулей
    differences = []
    for i in range(len(before)):
        d = before[i] - after[i]
        if d != 0:
            differences.append(d)

    if not differences:
        print("Все разности равны нулю. Изменений не обнаружено.")
        return 0

    # По требованию преподавателя сортируем со знаком
    sorted_differences = sorted(differences)

    ranks = []
    T_plus = 0
    T_minus = 0

    for d in differences:
        rank = (
            sorted_differences.index(d)
            + (sorted_differences.count(d) + 1) / 2
        )
        ranks.append(rank)

        if d > 0:
            T_plus += rank
        else:
            T_minus += rank

    T_emp = min(T_plus, T_minus)

    print("\nРазности:", differences)
    print("Ранги:", ranks)
    print("n =", len(differences))
    print("T+ =", T_plus)
    print("T− =", T_minus)
    print("T эмпирическое =", T_emp)
    print("T критическое =", T_crit)

    if T_emp <= T_crit:
        print("\nT эмп ≤ T крит.")
        print("H0 отвергаем. Данные поддерживают H1.")
        print("Изменения статистически значимы.")
    else:
        print("\nT эмп > T крит.")
        print("H0 не отвергаем. H1 не подтверждается.")
        print("Статистически значимых изменений не обнаружено.")

    return T_emp
```

```python
# Task 6. Evaluating the Effect of a Training Program.
before = [55, 62, 58, 70, 64, 59, 66, 61]
after = [60, 65, 63, 72, 68, 61, 70, 66]

T_emp = wilcoxon(before, after, T_crit=5)
```

```python
# Task 7. Comparing Blood Pressure Before and After an Intervention.
before = [150, 142, 155, 148, 160, 145, 152]
after = [145, 140, 150, 149, 154, 142, 146]

T_emp = wilcoxon(before, after, T_crit=3)
```
