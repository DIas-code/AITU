```python
import numpy as np

def anova(groups, F_critical):
    means_of_groups = []

    SUMall = 0
    N_all = 0

    for group in groups:
        means_of_groups.append(group.mean())
        SUMall += group.sum()
        N_all += len(group)

    print("Средние групп:", means_of_groups)

    overall_mean = SUMall / N_all
    print("Общее среднее:", overall_mean)

    SS_b = 0
    for group, mean in zip(groups, means_of_groups):
        SS_b += len(group) * (mean - overall_mean)**2
    print("SS_b:", SS_b)

    SS_w = 0
    for group, mean in zip(groups, means_of_groups):
        for x_i in group:
            SS_w += (x_i - mean)**2
    print("SS_w:", SS_w)

    MS_b = SS_b / (len(groups) - 1)
    print("MS_b:", MS_b)

    MS_w = SS_w / (N_all - len(groups))
    print("MS_w:", MS_w)

    F_emp = MS_b / MS_w
    print("F эмпирическое:", F_emp)
    print("F критическое:", F_critical)

    if F_emp > F_critical:
        print("Принимаем H1: средние статистически значимо различаются. Отвергаем H0.")
    else:
        print("H0 не отвергаем: значимые различия средних не обнаружены. H1 не подтверждена.")

    return F_emp
```

```python
from scipy import stats

def anova_library(groups, alpha=0.05):
    k = len(groups)
    N_all = sum(len(group) for group in groups)

    df_b = k - 1
    df_w = N_all - k

    F_emp, p_value = stats.f_oneway(*groups)
    F_critical = stats.f.ppf(1 - alpha, df_b, df_w)

    print("F эмпирическое:", F_emp)
    print("F критическое:", F_critical)
    print("p-value:", p_value)

    if p_value < alpha:
        print("Принимаем H1: средние статистически значимо различаются. Отвергаем H0.")
    else:
        print("H0 не отвергаем: значимые различия средних не обнаружены. H1 не подтверждена.")

    return F_emp, F_critical, p_value
```

## Task 2. Двухфакторная ANOVA

Исследователь хочет проверить, как тип удобрения
(Factor A: Fertilizer — A1, A2, A3) и уровень полива
(Factor B: Irrigation — Low, High) влияют на урожайность.

Для каждого сочетания удобрения и полива получено по 5 наблюдений.

### Данные

| Полив | A1 | A2 | A3 |
|---|---|---|---|
| Low | 18.2, 17.6, 19.0, 18.9, 17.8 | 16.0, 15.7, 16.4, 15.9, 15.5 | 17.3, 17.0, 17.8, 17.4, 17.1 |
| High | 22.5, 21.9, 23.1, 22.8, 22.0 | 19.1, 18.8, 19.5, 19.0, 18.6 | 20.9, 21.3, 21.0, 21.5, 21.1 |

### Гипотезы

**1. Основной эффект удобрения**

- H0: средняя урожайность одинакова для всех типов удобрения.
- H1: хотя бы у двух типов удобрения средняя урожайность различается.

**2. Основной эффект полива**

- H0: средняя урожайность одинакова при Low и High.
- H1: средняя урожайность различается при Low и High.

**3. Взаимодействие удобрения и полива**

- H0: влияние типа удобрения на урожайность не зависит от уровня полива.
- H1: влияние типа удобрения на урожайность зависит от уровня полива.

### Задача

При уровне значимости α = 0.05 проверить основной эффект удобрения,
основной эффект полива и взаимодействие факторов.

## Ручной код

```python
import numpy as np

A1_low = np.array([18.2, 17.6, 19.0, 18.9, 17.8])
A2_low = np.array([16.0, 15.7, 16.4, 15.9, 15.5])
A3_low = np.array([17.3, 17.0, 17.8, 17.4, 17.1])

A1_high = np.array([22.5, 21.9, 23.1, 22.8, 22.0])
A2_high = np.array([19.1, 18.8, 19.5, 19.0, 18.6])
A3_high = np.array([20.9, 21.3, 21.0, 21.5, 21.1])

# Строки — удобрения, столбцы — уровни полива
data = np.array([
    [A1_low,  A2_low,  A3_low],   # 11, 12, 13
    [A1_high, A2_high, A3_high]   # 21, 22, 23
])
b = 2
a = 3
n = 5
```

```python
mean_G = data.mean()
means_ij = data.mean(axis=2)
means_i = means_ij.mean(axis=1)
means_j = means_ij.mean(axis=0)

print("Общее среднее:", mean_G)
print("Средние ячеек 11, 12, 13 / 21, 22, 23:\n", means_ij)
print("Фактор B — средние строк, полив (Low, High):", means_i)
print("Фактор A — средние столбцов, удобрения (A1, A2, A3):", means_j)
```

```python
SS_B = 0

for mean in means_i:
    SS_B += a * n * (mean - mean_G)**2

MS_B = SS_B / b - 1
print(SS_B)
print(MS_B)

SS_A = 0

for mean in means_j:
    SS_A += b * n * (mean - mean_G)**2

MS_A = SS_A / a - 1
print(SS_A)
print(MS_A)

SS_error = 0

for i in range(b):          # строки — полив
    for j in range(a):      # столбцы — удобрения
        for value in data[i, j]:
            SS_error += (value - means_ij[i, j])**2

MS_error = SS_error / (data.size - a*b)
print(SS_error)
print(MS_error)

```

```python
# Средние квадраты A, B и ошибки
df_A = a - 1
df_B = b - 1
df_error = data.size - a * b

MS_A = SS_A / df_A
MS_B = SS_B / df_B
MS_error = SS_error / df_error

# Сумма квадратов между ячейками
SS_cells = 0

for i in range(b):          # строки — полив
    for j in range(a):      # столбцы — удобрения
        SS_cells += n * (means_ij[i, j] - mean_G)**2

# Взаимодействие факторов
SS_AB = SS_cells - SS_A - SS_B
df_AB = df_A * df_B
MS_AB = SS_AB / df_AB

# F-статистики
F_A = MS_A / MS_error
F_B = MS_B / MS_error
F_AB = MS_AB / MS_error

# Критические F, заданные вручную при alpha = 0.05
F_crit_a = 3.4
F_crit_b = 4.3
F_crit_ab = 3.4

# Фактор A — удобрение
print("Фактор A — удобрение")
print("F:", F_A)
print("F критическое:", F_crit_a)

if F_A > F_crit_a:
    print("Отвергаем H0: средние по удобрениям статистически значимо различаются.")
else:
    print("H0 не отвергаем: значимые различия средних по удобрениям не обнаружены.")

# Фактор B — полив
print("\nФактор B — полив")
print("F:", F_B)
print("F критическое:", F_crit_b)

if F_B > F_crit_b:
    print("Отвергаем H0: средние для Low и High статистически значимо различаются.")
else:
    print("H0 не отвергаем: значимые различия средних для Low и High не обнаружены.")

# Взаимодействие A и B
print("\nВзаимодействие удобрения и полива")
print("F:", F_AB)
print("F критическое:", F_crit_ab)

if F_AB > F_crit_ab:
    print("Отвергаем H0: влияние удобрения на урожайность зависит от уровня полива.")
else:
    print("H0 не отвергаем: статистически значимое взаимодействие не обнаружено.")
```

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols
from scipy import stats

# Данные
A1_low = np.array([18.2, 17.6, 19.0, 18.9, 17.8])
A2_low = np.array([16.0, 15.7, 16.4, 15.9, 15.5])
A3_low = np.array([17.3, 17.0, 17.8, 17.4, 17.1])

A1_high = np.array([22.5, 21.9, 23.1, 22.8, 22.0])
A2_high = np.array([19.1, 18.8, 19.5, 19.0, 18.6])
A3_high = np.array([20.9, 21.3, 21.0, 21.5, 21.1])

# Таблица: каждое наблюдение в отдельной строке
df = pd.DataFrame({
    "Fertilizer": (
        ["A1"] * 5 + ["A2"] * 5 + ["A3"] * 5
        + ["A1"] * 5 + ["A2"] * 5 + ["A3"] * 5
    ),
    "Irrigation": ["Low"] * 15 + ["High"] * 15,
    "Yield": np.concatenate([
        A1_low, A2_low, A3_low,
        A1_high, A2_high, A3_high
    ])
})

# Модель: оба фактора и их взаимодействие
model = ols(
    "Yield ~ C(Fertilizer) * C(Irrigation)",
    data=df
).fit()

# Таблица ANOVA
anova_table = sm.stats.anova_lm(model, typ=2)

anova_table["MS"] = anova_table["sum_sq"] / anova_table["df"]
anova_table = anova_table.rename(columns={
    "sum_sq": "SS",
    "F": "F_emp",
    "PR(>F)": "p-value"
})
anova_table = anova_table[["SS", "df", "MS", "F_emp", "p-value"]]

print("Таблица ANOVA:")
print(anova_table)

# Степени свободы
df_A = anova_table.loc["C(Fertilizer)", "df"]
df_B = anova_table.loc["C(Irrigation)", "df"]
df_AB = anova_table.loc["C(Fertilizer):C(Irrigation)", "df"]
df_error = anova_table.loc["Residual", "df"]

# Критические F через SciPy
alpha = 0.05

F_crit_a = stats.f.ppf(1 - alpha, df_A, df_error)
F_crit_b = stats.f.ppf(1 - alpha, df_B, df_error)
F_crit_ab = stats.f.ppf(1 - alpha, df_AB, df_error)

# Эмпирические F и p-value
F_A = anova_table.loc["C(Fertilizer)", "F_emp"]
F_B = anova_table.loc["C(Irrigation)", "F_emp"]
F_AB = anova_table.loc["C(Fertilizer):C(Irrigation)", "F_emp"]

p_A = anova_table.loc["C(Fertilizer)", "p-value"]
p_B = anova_table.loc["C(Irrigation)", "p-value"]
p_AB = anova_table.loc["C(Fertilizer):C(Irrigation)", "p-value"]

# Фактор A — удобрение
print("\nФактор A — удобрение")
print(f"F: {F_A:.4f}")
print(f"F критическое: {F_crit_a:.4f}")
print(f"p-value: {p_A:.6g}")

if F_A > F_crit_a:
    print("Отвергаем H0: средние по удобрениям статистически значимо различаются.")
else:
    print("H0 не отвергаем: значимые различия средних по удобрениям не обнаружены.")

# Фактор B — полив
print("\nФактор B — полив")
print(f"F: {F_B:.4f}")
print(f"F критическое: {F_crit_b:.4f}")
print(f"p-value: {p_B:.6g}")

if F_B > F_crit_b:
    print("Отвергаем H0: средние для Low и High статистически значимо различаются.")
else:
    print("H0 не отвергаем: значимые различия средних для Low и High не обнаружены.")

# Взаимодействие факторов
print("\nВзаимодействие удобрения и полива")
print(f"F: {F_AB:.4f}")
print(f"F критическое: {F_crit_ab:.4f}")
print(f"p-value: {p_AB:.6g}")

if F_AB > F_crit_ab:
    print("Отвергаем H0: влияние удобрения на урожайность зависит от уровня полива.")
else:
    print("H0 не отвергаем: статистически значимое взаимодействие не обнаружено.")
```
