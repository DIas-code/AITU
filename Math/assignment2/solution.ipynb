```python
import numpy as np

def anova(groups, F_critical):
    means_of_groups = []

    SUMall = 0
    N_all = 0

    for group in groups:
        means_of_groups.append(group.mean())
        SUMall += group.sum()
        N_all += len(group)

    print("Средние групп:", means_of_groups)

    overall_mean = SUMall / N_all
    print("Общее среднее:", overall_mean)

    SS_b = 0
    for group, mean in zip(groups, means_of_groups):
        SS_b += len(group) * (mean - overall_mean)**2
    print("SS_b:", SS_b)

    SS_w = 0
    for group, mean in zip(groups, means_of_groups):
        for x_i in group:
            SS_w += (x_i - mean)**2
    print("SS_w:", SS_w)

    MS_b = SS_b / (len(groups) - 1)
    print("MS_b:", MS_b)

    MS_w = SS_w / (N_all - len(groups))
    print("MS_w:", MS_w)

    F_emp = MS_b / MS_w
    print("F эмпирическое:", F_emp)
    print("F критическое:", F_critical)

    if F_emp > F_critical:
        print("Принимаем H1: средние статистически значимо различаются. Отвергаем H0.")
    else:
        print("H0 не отвергаем: значимые различия средних не обнаружены. H1 не подтверждена.")

    return F_emp
```

```python

tv = np.array([6.525, 7.069, 7.065, 7.359, 7.636, 7.929])
radio = np.array([10.418, 8.016, 7.608, 8.175, 8.335, 8.397])
newspapers = np.array([6.062, 8.609, 7.143, 6.923, 5.857])
internet = np.array([5.086, 8.198, 7.009, 7.972])
```

```python
groups_1 = [tv, radio, newspapers, internet]
F_critical = 3.20
anova(groups_1, F_critical)
```

```python
from scipy import stats

def anova_library(groups, alpha=0.05):
    k = len(groups)
    N_all = sum(len(group) for group in groups)

    df_b = k - 1
    df_w = N_all - k

    F_emp, p_value = stats.f_oneway(*groups)
    F_critical = stats.f.ppf(1 - alpha, df_b, df_w)

    print("F эмпирическое:", F_emp)
    print("F критическое:", F_critical)
    print("p-value:", p_value)

    if p_value < alpha:
        print("Принимаем H1: средние статистически значимо различаются. Отвергаем H0.")
    else:
        print("H0 не отвергаем: значимые различия средних не обнаружены. H1 не подтверждена.")

    return F_emp, F_critical, p_value
```
